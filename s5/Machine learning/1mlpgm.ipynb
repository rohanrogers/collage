```python
#cell 5

#training
#1. load the data
import pandas as pd
file=pd.read_csv("/content/irisexcel.csv")
x=file[["sepal_length","sepal_width","petal_length","petal_width"]]
y=file["species"]
#crossvalidation
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8)
#2. load the model
from sklearn.linear_model import Perceptron
model=Perceptron()
#3. fit the data into model
model.fit(xtrain,ytrain)

#validation
op=model.predict(xtest)
from sklearn.metrics import accuracy_score
score=accuracy_score(ytest,op)
print("accuracy =", score*100)

#testing......predict
op2=model.predict([[2.5,1.5,3.5,5.5]])
print(op2)
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(ytest,op)
print(cm)
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cmd = ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=model.classes_)
cmd.plot()
```
